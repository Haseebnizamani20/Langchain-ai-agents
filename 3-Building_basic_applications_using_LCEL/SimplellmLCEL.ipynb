```python
from langchain_groq import ChatGroq
model=ChatGroq(model="llama-3.3-70b-versatile",groq_api_key=groq_api_key)
```

```python
from langchain_core.messages import HumanMessage,SystemMessage
# sysem massage means how the system should behave
# human message means what the user is asking       

messages = [
    SystemMessage(content="Trasnalte the followwing from English to French"),
    HumanMessage(content="Hellow how are you"),
]
result=model.invoke(messages)


```

```python
result
from langchain_core.output_parsers import StrOutputParser
parser=StrOutputParser()
parser.invoke(result)


```

```python
# USING LCEL -CHIAN THE COMPONENTS
chain=model | parser
chain.invoke(messages)


```

```python
# prompt template


from langchain_core.prompts import ChatPromptTemplate 
generic_template="translate the following {language} : "

prompt =ChatPromptTemplate.from_messages(
    [("system",generic_template),("user","{text}")]
    )
```

```python
prompt.invoke({"language":"urdu","text":"Hellow how are you"})
```

```python
chain=prompt|model|parser
chain.invoke({"language":"sindhi","text":"Hellow how are you"})
```
